```python
texts = [ "Free money now!!!", "Hey, are we still meeting tomorrow?", "Congratulations! You've won a free gift card!", "Reminder about your appointment next week", "Urgent! Reply to claim your prize!", "Let's catch up later tonight.", "Exclusive offer for you!", "This is not a spam message." ] 
```

```python
 labels = [1, 0, 1, 0, 1, 0, 1, 0]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vec = CountVectorizer()
X = vec.fit_transform(texts) # convert test to features vectors
```

```python
pd.set_option("display.max_columns", None)
cols = vec.get_feature_names_out()
pd.DataFrame(X.toarray(), columns=cols)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.2, random_state=42)
```

```python
nb_classifier = MultinomialNB()
```

```python
nb_classifier.fit(X_train, y_train)
```

```python
y_pred_train = nb_classifier.predict(X_train)
y_pred_test = nb_classifier.predict(X_test)
```
